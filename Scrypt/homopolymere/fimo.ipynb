```python
for i in range(20,21):
    path_global = '../output/snakemake'
    path = f'{path_global}/output/fimo/{i}/'
    !fimo -oc {path_global}/output/fimo/{i}/ {path_global}/output/{i}/meme/meme.txt  ../data/Fasta/genome.fasta


    input_file = f"{path}fimo.gff"
    output_file = f"{path}fimo_filtered.gff"
    
    with open(input_file, "r") as infile, open(output_file, "w") as outfile:
        for line in infile:
            if line.startswith("#") or line.strip() == "":
                continue  # ignorer commentaires ou lignes vides
    
            fields = line.strip().split("\t")
            if len(fields) < 9:
                continue  # ligne mal formée
    
            attributes = fields[8]
            attr_dict = dict(attr.split("=") for attr in attributes.strip(";").split(";") if "=" in attr)
    
            qvalue_str = attr_dict.get("qvalue")
            if qvalue_str is not None:
                try:
                    qvalue = float(qvalue_str)
                    if qvalue < 10**-10:
                        outfile.write(line)
                except ValueError:
                    continue  # valeur invalide, on ignore
```
