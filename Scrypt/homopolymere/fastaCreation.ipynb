```python
def load_fasta(fasta_path, pos,name):
    f = FastA(fasta_path)
    sequences = []
    i = 1
    for maseq in f:

        print(maseq.name)

        try:
            num_pos = int(maseq.name)
        except:
            num_pos = i
        
        if num_pos in pos:
            
            pos_start, pos_end, pos_length = pos[num_pos]
        else:
            pos_start = 0
            pos_end = 0
            pos_length = 0
            print('bas')

        if pos_start != 0:
            seq = maseq.sequence.upper()
            seq = seq[pos_start:pos_end]
                
            
            sequences.append( SeqRecord(Seq(str(seq)), id=maseq.name, description=maseq.name))
        i = i+1
    with open(f'../output/estimation/sequences/especes/{name}_centromeres.fasta', "w") as output_handle:
        SeqIO.write(sequences, output_handle, "fasta")
        
#load_fasta("../data/Fasta/genomes/Linfantum/TriTrypDB-68_LinfantumJPCM5_Genome.fasta",infantum,'Linfantum')
#load_fasta("../data/Fasta/TriTrypDB-68_LmajorFriedlin_Genome.fasta",major,'Lmajor')
#load_fasta("../data/Fasta/genome.fasta",donovani,'Ldonovani')
#load_fasta("../data/Fasta/genomes/Laethiopica/ragtag.scaffold.fasta",aethiopica,'Laethiopica')
#load_fasta("../data/Fasta/genomes/LamazonensisPH8/ragtag.scaffold.fasta",amazonensis,'Lamazonensis')
#load_fasta("../data/Fasta/genomes/Larabica/ragtag.scaffold.fasta",arabica,'Larabica')
#load_fasta("../data/Fasta/genomes/Lmexicana/ragtag.scaffold.fasta",mexicana,'Lmexicana')
load_fasta("../data/Fasta/genomes/Ltropica/LtropicaRef_genome.fasta",tropica,'Ltropica')
#load_fasta("../data/Fasta/genomes/Ltarentolae/ragtag.scaffold.fasta",tarentolae,'Ltarentolae')
#load_fasta("../data/Fasta/genomes/Lbraziliensis/ragtag.scaffold.fasta",braziliensis,'Lbraziliensis')
load_fasta("../data/Fasta/genomes/Lpanamensis/TriTrypDB-68_LpanamensisMHOMPA94PSC1_Genome.fasta",panamensis,'Lpanamensis')
#load_fasta("../data/Fasta/genomes/Lorientalis/ragtag.scaffold.fasta",orientalis,'Lorientalis')
#load_fasta("../data/Fasta/genomes/Lmartiniquensis/ragtag.scaffold.fasta",braziliensis,'Lmartiniquensis')

```

```python

def delete_fasta(name, ids_to_remove):

    input_file = f"../output/estimation/sequences/especes/L{name}_centromeres.fasta"

    # Lire, filtrer, puis écrire
    with open(input_file, "r") as infile:
        records = SeqIO.parse(infile, "fasta")
        filtered_records = [record for record in records if record.id not in ids_to_remove]
    
    # Sauvegarder le nouveau fichier FASTA
    with open(input_file, "w") as outfile:
        SeqIO.write(filtered_records, outfile, "fasta")



ids_to_remove = {"1", "29"}
delete_fasta('aethiopica',ids_to_remove)


ids_to_remove = {"18", "29"}
delete_fasta('arabica',ids_to_remove)


ids_to_remove = {"18", "29","25"}
delete_fasta('tropica',ids_to_remove)



ids_to_remove = {"20", "28"}
delete_fasta('orientalis',ids_to_remove)



ids_to_remove = {"1"}
delete_fasta('martiniquensis',ids_to_remove)

ids_to_remove = {"11","24","35","36"}
delete_fasta('panamensis',ids_to_remove)


```
