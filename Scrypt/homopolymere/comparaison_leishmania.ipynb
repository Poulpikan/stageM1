```python

# --- préparation ----------------------------------------------------------
newdef = df.copy()
#del newdef['delta']
del newdef['papier']


# --- copie profonde (plus sûr) -------------------------------------------
df_orientalis = newdef['orientalis'].copy()

# sauvegarde des valeurs à permuter
temp33 = df_orientalis.loc['Chr33'].copy()
temp34 = df_orientalis.loc['Chr34'].copy()

# permutation
df_orientalis.loc['Chr33'] = df_orientalis.loc['Chr35']
df_orientalis.loc['Chr34'] = df_orientalis.loc['Chr36']
df_orientalis.loc['Chr35'] = temp33
df_orientalis.loc['Chr36'] = temp34

# on remet la version modifiée dans le dictionnaire
newdef['orientalis'] = df_orientalis


g = sns.clustermap(
    newdef.fillna(3000),
    figsize=(7, 5),
    row_cluster=False,
    dendrogram_ratio=(.1, .2),
    cbar_pos=(0, .2, .03, .4)
)
g.figure.savefig("mon_clustermap.png", dpi=150, bbox_inches='tight')  # ou .pdf, .svg, etc.
```
